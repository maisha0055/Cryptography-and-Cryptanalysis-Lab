```python
import os
import sys
import binascii

from cryptography.hazmat.primitives import cmac
from cryptography.hazmat.decrepit.ciphers import algorithms

message = "I met an interesting turtle while the song on the radio blasted away"
message_bytes = message.encode()  # Convert string to bytes
key = b'\x01\xd8i\xa1^0\x9a<\x0f\xf0\r\xc1\xdd\xd5\x89\xa6'
# You code here. HINT: use CMAC, TripleDES

c = cmac.CMAC(algorithms.TripleDES(key))

c.update(message_bytes)

signature = c.finalize()

print(f"Message: {message}")
print(f"CMAC signature: {binascii.b2a_hex(signature).decode()}")
```

```python
#verify the given information here
given_signature = 'ba4ecb8db45c6ae0'
if signature.hex() == given_signature:
    print("Valid")
else:
    print("Invalid")
```

```python
def A51(X, Y, Z, n):
    if isinstance(X, str): X = [int(b) for b in X]
    if isinstance(Y, str): Y = [int(b) for b in Y]
    if isinstance(Z, str): Z = [int(b) for b in Z]

    X = list(X)
    Y = list(Y)
    Z = list(Z)

    keystream = []

    for _ in range(n):
        x_clock = X[8]
        y_clock = Y[10]
        z_clock = Z[10]
        majority = 1 if (x_clock + y_clock + z_clock) >= 2 else 0

        if x_clock == majority:
            fb = X[18] ^ X[17] ^ X[16] ^ X[13]
            X = [fb] + X[:-1]

        if y_clock == majority:
            fb = Y[21] ^ Y[20]
            Y = [fb] + Y[:-1]

        if z_clock == majority:
            fb = Z[22] ^ Z[21] ^ Z[20] ^ Z[7]
            Z = [fb] + Z[:-1]

        out = X[18] ^ Y[21] ^ Z[22]
        keystream.append(out)

    return keystream


def encrypt_A51(plaintext, X, Y, Z):
    n = len(plaintext) * 8
    ks = A51(X, Y, Z, n)

    pt_bits = []
    for ch in plaintext:
        byte_val = ord(ch)
        for i in range(7, -1, -1):
            pt_bits.append((byte_val >> i) & 1)

    ct_bits = [p ^ k for p, k in zip(pt_bits, ks)]

    ct_bytes = []
    for i in range(0, len(ct_bits), 8):
        byte = 0
        for bit in ct_bits[i:i+8]:
            byte = (byte << 1) | bit
        ct_bytes.append(byte)

    keystream_str = ''.join(map(str, ks))
    pt_binary = ''.join(map(str, pt_bits))
    ct_binary = ''.join(map(str, ct_bits))
    ct_hex = ''.join(f'{b:02x}' for b in ct_bytes)

    return keystream_str, pt_binary, ct_binary, ct_hex


def decrypt_A51(ciphertext_hex, X, Y, Z):
    ct_bytes = bytes.fromhex(ciphertext_hex)
    n = len(ct_bytes) * 8
    ks = A51(X, Y, Z, n)

    ct_bits = []
    for byte in ct_bytes:
        for i in range(7, -1, -1):
            ct_bits.append((byte >> i) & 1)

    pt_bits = [c ^ k for c, k in zip(ct_bits, ks)]

    result = []
    for i in range(0, len(pt_bits), 8):
        byte = 0
        for bit in pt_bits[i:i+8]:
            byte = (byte << 1) | bit
        result.append(chr(byte))

    return ''.join(result)


print("A5/1 functions defined successfully.")
```

```python
X = "1110001100101001011"   # 19
Y = "0011000000010000001101" # 22
Z = "10011101101111001001110" # 23

plaintexts = [
    "It is alive",
    "Snap out of it",
    "I am as mad as hell and I am not going to take this anymore",
    "Bond James Bond",
    "Love means never having to say you're sorry"
]

print(f"Initial states: X={X}  Y={Y}  Z={Z}")

results = []
for pt in plaintexts:
    ks_str, pt_bin, ct_bin, ct_hex = encrypt_A51(pt, X, Y, Z)
    decrypted = decrypt_A51(ct_hex, X, Y, Z)
    results.append((pt, ks_str, pt_bin, ct_bin, ct_hex, decrypted))

    print(f"\nPlaintext      : {pt}")
    print(f"Key stream     : {ks_str}")
    print(f"PT (binary)    : {pt_bin}")
    print(f"CT (binary)    : {ct_bin}")
    print(f"CT (hex)       : {ct_hex}")
    print(f"Decrypted      : {decrypted}")
    print(" ")
```
